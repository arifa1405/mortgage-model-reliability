Check the existing source code rather than redefining the original target. The first scheduled payment month is month 0; month 23 is the last included month. Terminal events are timed by reporting month.

## 2. Record the candidate Fannie event definitions

```python
ADVERSE = {'02','03','09','15'}
NON_ADVERSE = {'01','16','96'}
REPURCHASE = {'06'}
code_map = pd.DataFrame([
    ('01','Prepayment or maturity','non-adverse candidate'),
    ('02','Third-party sale','adverse candidate'),
    ('03','Short sale','adverse candidate'),
    ('06','Repurchase','separate eligibility review'),
    ('09','Deed-in-lieu or REO disposition','adverse candidate'),
    ('15','Non-performing note sale','adverse candidate'),
    ('16','Reperforming note sale','non-adverse candidate'),
    ('96','Non-credit removal','non-adverse candidate')], columns=['code','working_definition','candidate_treatment'])
print(code_map.to_string(index=False))
```

## 3. Recreate the reviewed sample
Scan the file using the same hash, physical positions and memory guard as Notebook 14B. Stop if counts differ from the reviewed run.

```python
SIZE = SOURCE.stat().st_size
WIDTH = 113
assert SIZE == 6928164686, 'Source differs from reviewed file.'
POSITIONS = {'loan_id':2,'report':3,'orig':14,'first_payment':15,'age':16,'delinq':40,'payment_history':41,'modification':42,'zero_code':44,'zero_month':45}
SAMPLE_MODULUS = 100
MAX_SAMPLE_ROWS = 1_000_000
SOURCE_SIGNATURE = (SOURCE.stat().st_size, SOURCE.stat().st_mtime_ns)
@lru_cache(maxsize=100_000)
def selected(loan_id):
    return int.from_bytes(hashlib.blake2b(loan_id,digest_size=8,person=b'Fannie14v1').digest(),'big') % SAMPLE_MODULUS == 0
def scan_source(path, expected_width, progress_every=2_000_000):
    counts = Counter()
    sampled = []
    widths_seen = Counter()
    id_pattern = re.compile(rb'[A-Za-z0-9]{1,12}')
    indexes = [p-1 for p in POSITIONS.values()]
    with path.open('rb') as stream:
        for line_number, raw in enumerate(stream, 1):
            parts = raw.rstrip(b'\r\n').split(b'|')
            if line_number == 1:
                parts[0] = parts[0].removeprefix(b'\xef\xbb\xbf')
            counts['physical_rows'] += 1
            widths_seen[len(parts)] += 1
            if len(parts) != expected_width:
                counts['wrong_width'] += 1
                continue
            loan_id = parts[1].strip()
            if not id_pattern.fullmatch(loan_id):
                counts['invalid_loan_id'] += 1
                continue
            counts['valid_id_rows'] += 1
            if selected(loan_id):
                sampled.append(tuple(parts[i].strip().decode('ascii') for i in indexes))
                if len(sampled) > MAX_SAMPLE_ROWS:
                    raise RuntimeError('Sample exceeds memory guard; increase SAMPLE_MODULUS and rerun the entire scan.')
            if progress_every and line_number % progress_every == 0:
                print(f'{line_number:,} rows scanned; {len(sampled):,} sampled rows')
    return pd.DataFrame(sampled, columns=list(POSITIONS)), counts, widths_seen

history, scan_counts, full_widths = scan_source(SOURCE, WIDTH)
assert SOURCE_SIGNATURE == (SOURCE.stat().st_size, SOURCE.stat().st_mtime_ns), 'Source changed during scan.'
print(dict(scan_counts), dict(full_widths))
assert not scan_counts['wrong_width'] and not scan_counts['invalid_loan_id'], 'Malformed records found; review before continuing.'
assert not history.empty, 'No sampled rows'
print('Sample rows:', len(history), '| Sample distinct loans:', history.loan_id.nunique())
assert scan_counts['physical_rows'] == 23563399
assert len(history) == 245552 and history.loan_id.nunique() == 4749, 'Sample differs from reviewed Stage 2.' 
```

## 4. Check dates and observed codes
Keep unknown statuses separate. Check all dates and duplicate loan-month records before comparing candidate rules.

```python
def month_index(series):
    valid = series.str.fullmatch(r'(0[1-9]|1[0-2])(19|20)\d{2}')
    safe = series.where(valid)
    return (pd.to_numeric(safe.str[2:]) * 12 + pd.to_numeric(safe.str[:2]) - 1).astype('Int64')

def month_text(i):
    if pd.isna(i): return 'missing'
    return f'{int(i)//12:04d}-{int(i)%12+1:02d}'

for name in ['report','orig','first_payment','zero_month']:
    history[name+'_idx'] = month_index(history[name])
    print(name, 'blank:', int(history[name].eq('').sum()), 'invalid nonblank:', int((history[name].ne('') & history[name+'_idx'].isna()).sum()))

history['numeric_delinq'] = pd.to_numeric(history.delinq.where(history.delinq.str.fullmatch(r'\d{1,2}')), errors='coerce')
valid_report = history[history.report_idx.notna()]
# Count duplicates without discarding conflicting rows.
duplicate_counts = valid_report.groupby(['loan_id','report_idx']).size()
duplicate_groups = duplicate_counts[duplicate_counts > 1]
print('Duplicate loan-month groups:', len(duplicate_groups), '| Extra records:', int((duplicate_groups-1).sum()))
print('Exact repeated selected-field records:', int(history[list(POSITIONS)].duplicated().sum()))
print('Delinquency codes (row counts):')
print(history.delinq.replace('', '<blank>').value_counts().to_string())
print('Zero balance codes (row counts):')
print(history.zero_code.replace('', '<blank>').value_counts().to_string())
assert len(duplicate_groups)==0
assert history.report_idx.notna().all()
known_codes = ADVERSE | NON_ADVERSE | REPURCHASE
unexpected = sorted(set(history.zero_code) - known_codes - {''})
assert not unexpected, f'Review unexpected terminal codes: {unexpected}'
```

## 5. Compare loan-level timing and eligibility
Do not choose a date when records disagree. Compare reporting and effective terminal dates, distinguish unknowns at termination from active-month unknowns, and count how stricter observation rules would change eligibility. No final target column is created.

```python
audit = []
for loan_id, g in history.groupby('loan_id', sort=False):
    g = g.sort_values('report_idx')
    orig = g.orig_idx.dropna().unique()
    fp = g.first_payment_idx.dropna().unique()
    row = dict(orig_conflict=len(orig)!=1, fp_conflict=len(fp)!=1)
    if len(orig)!=1 or len(fp)!=1 or g.orig_idx.isna().any() or g.first_payment_idx.isna().any():
        row['date_resolved'] = False
        audit.append(row)
        continue
    row['date_resolved'] = True
    start = int(fp[0]); end = start+23
    row['origination_quarter'] = str(pd.Period(year=int(orig[0])//12, month=int(orig[0])%12+1, freq='M').asfreq('Q'))
    w = g[g.report_idx.between(start,end)]
    terminal = g[g.zero_code.ne('')]
    tr = terminal[terminal.report_idx.between(start,end)]
    te = terminal[terminal.zero_month_idx.between(start,end).fillna(False)]
    row['left_truncated'] = int(g.report_idx.min()) > start
    row['followup_complete'] = int(g.report_idx.max()) >= end or not tr.empty
    row['candidate_freddie_eligible'] = not row['left_truncated'] and row['followup_complete']
    row['terminal_report_in_window'] = not tr.empty
    row['terminal_effective_in_window'] = not te.empty
    row['terminal_window_disagrees'] = (not tr.empty) != (not te.empty)
    row['terminal_date_differs'] = bool((terminal.report_idx != terminal.zero_month_idx).fillna(True).any())
    row['candidate_adverse_report'] = bool(tr.zero_code.isin(ADVERSE).any())
    row['candidate_adverse_effective'] = bool(te.zero_code.isin(ADVERSE).any())
    row['candidate_serious_numeric'] = bool(w.numeric_delinq.ge(3).any())
    row['repurchase_in_window'] = bool(tr.zero_code.isin(REPURCHASE).any())
    unknown = w.numeric_delinq.isna()
    row['unknown_any'] = bool(unknown.any())
    row['unknown_terminal_row'] = bool((unknown & w.zero_code.ne('')).any())
    row['unknown_active_row'] = bool((unknown & w.zero_code.eq('')).any())
    last_required = min(end, int(tr.report_idx.min())) if not tr.empty else end
    required = set(range(start,last_required+1))
    observed = set(w.report_idx.astype(int))
    row['missing_required_months'] = len(required-observed)
    row['candidate_strict_eligible'] = row['candidate_freddie_eligible'] and not row['missing_required_months'] and not row['unknown_active_row'] and not row['repurchase_in_window']
    row['terminal_code_in_window'] = '|'.join(sorted(set(tr.zero_code))) or '<none>'
    audit.append(row)
audit = pd.DataFrame(audit)
assert len(audit)==4749
resolved = audit[audit.date_resolved].copy()
print('Resolved dates:', len(resolved), 'Unresolved:', len(audit)-len(resolved))
```

## 6. Summarize diagnostic comparisons
Count unresolved loans separately. Compare candidate eligibility and event flags without treating unknown statuses as negative outcomes. No loan IDs are exported.

```python
checks = []
for name in ['orig_conflict','fp_conflict','date_resolved']:
    checks.append({'check':name,'count':int(audit[name].eq(True).sum()),'denominator':len(audit)})
for name in resolved.columns:
    if name not in ['orig_conflict','fp_conflict','date_resolved','origination_quarter','terminal_code_in_window','missing_required_months']:
        checks.append({'check':name,'count':int(resolved[name].eq(True).sum()),'denominator':len(resolved)})
checks = pd.DataFrame(checks)
print(checks.to_string(index=False))
comparison = resolved.groupby(['candidate_freddie_eligible','candidate_strict_eligible','candidate_serious_numeric','candidate_adverse_report','unknown_active_row','unknown_terminal_row','repurchase_in_window'],dropna=False).size().reset_index(name='loans')
terminal_counts = resolved.groupby(['terminal_code_in_window','terminal_window_disagrees','terminal_date_differs'],dropna=False).size().reset_index(name='loans')
cohort_counts = resolved.groupby(['origination_quarter','candidate_freddie_eligible','candidate_strict_eligible']).size().reset_index(name='loans')
conflict_patterns = audit[~audit.date_resolved].groupby(['orig_conflict','fp_conflict']).size().reset_index(name='loans')
coverage = resolved.groupby('missing_required_months').size().reset_index(name='loans')
assert comparison.loans.sum()==terminal_counts.loans.sum()==cohort_counts.loans.sum()==coverage.loans.sum()==len(resolved)
print(comparison.to_string(index=False))
```

## Stage 3D — Construct diagnostic-sample targets
The primary rule mirrors Freddie eligibility; the sensitivity rule excludes repurchases, unknown active statuses and missing required months. Exclude unresolved dates under both. These are working labels for the reviewed sample, not release-certified or final external-evaluation data. Unknown terminal status is not imputed to zero. The July 2024 mapping remains provisional until release-specific definitions are confirmed.

```python
def build_target(g):
    orig = g.orig_idx.dropna().unique()
    fp = g.first_payment_idx.dropna().unique()
    row = dict(date_resolved=False, primary_eligible=False, strict_eligible=False,
               serious=False, adverse=False, repurchase=False,
               unknown_active=False, missing_months=None, quarter=None,
               reason='unresolved_dates', first_serious_month=None,
               first_adverse_month=None)
    if len(orig)!=1 or len(fp)!=1 or g.orig_idx.isna().any() or g.first_payment_idx.isna().any():
        return row
    start = int(fp[0]); end = start+23
    row['date_resolved'] = True
    row['quarter'] = str(pd.Period(year=int(orig[0])//12, month=int(orig[0])%12+1, freq='M').asfreq('Q'))
    w = g[g.report_idx.between(start,end)]
    tr = w[w.zero_code.ne('')]
    serious = w.numeric_delinq.ge(3)
    adverse = w.zero_code.isin(ADVERSE)
    left = int(g.report_idx.min()) > start
    complete = int(g.report_idx.max()) >= end or not tr.empty
    endpoint = min(end,int(tr.report_idx.min())) if not tr.empty else end
    missing = len(set(range(start,endpoint+1))-set(w.report_idx.astype(int)))
    unknown_active = bool((w.numeric_delinq.isna() & w.zero_code.eq('')).any())
    repurchase = bool(tr.zero_code.isin(REPURCHASE).any())
    primary = not left and complete
    strict = primary and missing==0 and not unknown_active and not repurchase
    row.update(primary_eligible=primary, strict_eligible=strict,
               serious=bool(serious.any()), adverse=bool(adverse.any()),
               repurchase=repurchase, unknown_active=unknown_active, missing_months=missing,
               reason='left_truncated' if left else ('incomplete_followup' if not complete else 'eligible'),
               first_serious_month=month_text(w.loc[serious,'report_idx'].min()) if serious.any() else None,
               first_adverse_month=month_text(w.loc[adverse,'report_idx'].min()) if adverse.any() else None)
    return row

# Test the rule with small artificial histories before using the sample.
def example(months, events=None, terminals=None, conflict=False):
    start=2018*12+2
    events=events or {}; terminals=terminals or {}
    return pd.DataFrame(dict(orig_idx=[start-2+(int(conflict) if i==1 else 0) for i in range(len(months))],
        first_payment_idx=start, report_idx=[start+m for m in months],
        numeric_delinq=[events.get(m,0) for m in months], zero_code=[terminals.get(m,'') for m in months]))

tests=[
 ('complete_no_event',example(list(range(24))),True,True,False),
 ('month_0_event',example(list(range(24)),{0:3}),True,True,True),
 ('month_23_event',example(list(range(24)),{23:3}),True,True,True),
 ('month_24_excluded',example(list(range(25)),{24:3}),True,True,False),
 ('prepayment_unknown',example(list(range(7)),{6:float('nan')},{6:'01'}),True,True,False),
 ('adverse_terminal',example(list(range(7)),{6:float('nan')},{6:'09'}),True,True,True),
 ('repurchase',example(list(range(7)),{6:float('nan')},{6:'06'}),True,False,False),
 ('late_start',example(list(range(1,24))),False,False,False),
 ('incomplete',example(list(range(12))),False,False,False),
 ('gap',example([m for m in range(24) if m!=5]),True,False,False),
 ('unknown_active',example(list(range(24)),{5:float('nan')}),True,False,False),
 ('conflict',example(list(range(24)),conflict=True),False,False,False)]
results=[]
for name,g,p,s,e in tests:
    r=build_target(g)
    passed=(r['primary_eligible'],r['strict_eligible'],r['serious'] or r['adverse'])==(p,s,e)
    results.append(dict(test=name,passed=passed))
    assert passed,name
boundary_tests=pd.DataFrame(results)
print(boundary_tests.to_string(index=False))
```

## 8. Create one row per sampled loan
Keep identifiers and labels in memory only. Ineligible targets remain missing. Compare both policies on exactly the same loans.

```python
rows=[]
for loan_id,g in history.groupby('loan_id',sort=False):
    rows.append(dict(loan_id=loan_id,**build_target(g)))
targets=pd.DataFrame(rows)
assert len(targets)==4749 and targets.loan_id.is_unique
for policy in ['primary','strict']:
    eligible=targets[policy+'_eligible']
    targets[policy+'_90_plus']=pd.Series(pd.NA,index=targets.index,dtype='Int8')
    targets[policy+'_composite']=pd.Series(pd.NA,index=targets.index,dtype='Int8')
    targets.loc[eligible,policy+'_90_plus']=targets.loc[eligible,'serious'].astype('int8')
    targets.loc[eligible,policy+'_composite']=(targets.loc[eligible,'serious'] | targets.loc[eligible,'adverse']).astype('int8')
    assert targets[policy+'_composite'].notna().equals(eligible)
assert not (targets.strict_eligible & ~targets.primary_eligible).any()
assert (targets.loc[targets.strict_eligible,'primary_composite']==targets.loc[targets.strict_eligible,'strict_composite']).all()
# The earlier audit iterated over the same group order; verify row-level flags.
assert len(audit)==len(targets)
assert targets.date_resolved.equals(audit.date_resolved)
for new,old in [('primary_eligible','candidate_freddie_eligible'),('strict_eligible','candidate_strict_eligible'),('serious','candidate_serious_numeric'),('adverse','candidate_adverse_report')]:
    assert targets[new].eq(audit[old].eq(True)).all(),new
print('Loan-level diagnostic reconciliation passed.')
```

## Stage 3E — Validate and export aggregate results
Check the totals against the previously reviewed ZIP. Report actual origination quarters separately; the 2017 loans are not part of the intended out-of-time period.

```python
summary=[]; cohorts=[]
for policy,expected in [('primary',4491),('strict',4488)]:
    eligible=targets[targets[policy+'_eligible']]
    n=len(eligible); events=int(eligible[policy+'_composite'].sum())
    assert n==expected and events==33
    assert int(eligible.serious.sum())==33 and int(eligible.adverse.sum())==3
    summary.append(dict(policy=policy,sample_loans=len(targets),eligible=n,ineligible=len(targets)-n,
                        serious=int(eligible.serious.sum()),adverse=int(eligible.adverse.sum()),
                        both=int((eligible.serious & eligible.adverse).sum()),composite=events,event_rate=events/n))
    for quarter,g in targets.groupby('quarter',dropna=False):
        e=g[g[policy+'_eligible']]; positives=int(e[policy+'_composite'].sum())
        cohorts.append(dict(policy=policy,origination_quarter=quarter,loans=len(g),eligible=len(e),events=positives,event_rate=positives/len(e) if len(e) else None))
summary=pd.DataFrame(summary); cohorts=pd.DataFrame(cohorts)
assert targets.date_resolved.sum()==4744
assert int((targets.primary_eligible & ~targets.strict_eligible).sum())==3
assert targets.loc[targets.primary_eligible & ~targets.strict_eligible,'repurchase'].all()
assert SOURCE_SIGNATURE==(SOURCE.stat().st_size,SOURCE.stat().st_mtime_ns)
print(summary.to_string(index=False)); print(cohorts.to_string(index=False))
reasons=targets.groupby(['reason','primary_eligible','strict_eligible'],dropna=False).size().reset_index(name='loans')
composition=targets.groupby(['primary_eligible','strict_eligible','serious','adverse','repurchase','unknown_active'],dropna=False).size().reset_index(name='loans')
assert reasons.loans.sum()==composition.loans.sum()==len(targets)
review=pd.DataFrame([dict(check='reviewed_sample_totals_and_events',passed=True),dict(check='row_level_diagnostic_flags',passed=True),dict(check='strict_subset_and_label_agreement',passed=True),dict(check='boundary_tests',passed=bool(boundary_tests.passed.all()))])
final_out=out/'stage3de'; final_out.mkdir(exist_ok=False)
for name,table in {'target_summary.csv':summary,'cohort_target_summary.csv':cohorts,'eligibility_reasons.csv':reasons,'event_composition.csv':composition,'boundary_tests.csv':boundary_tests,'reconciliation_checks.csv':review}.items():
    table.to_csv(final_out/name,index=False)
final_meta=dict(meta,stage='3D–3E',diagnostic_sample_targets_constructed=True,final_target_constructed=False,release_vintage='unverified',primary_policy='Freddie-aligned working mapping',sensitivity_policy='strict coverage, no active unknowns or repurchases',prior_review_sha256_expected='08475a8b89f1d1b21ef522ea124ea23c8dd044e1c38cae587001d462db31492d',source_code_matches_review=freddie_sha=='08475a8b89f1d1b21ef522ea124ea23c8dd044e1c38cae587001d462db31492d')
assert final_meta['source_code_matches_review'], 'Freddie source changed since review.'
(final_out/'stage3de_metadata.json').write_text(json.dumps(final_meta,indent=2),encoding='utf-8')
final_zip=final_out/'notebook_14c_stage3de_review.zip'
with zipfile.ZipFile(final_zip,'w',compression=zipfile.ZIP_DEFLATED) as z:
    for f in sorted(final_out.iterdir()):
        if f!=final_zip: z.write(f,f.name)
print('Review ZIP:',final_zip)
```
